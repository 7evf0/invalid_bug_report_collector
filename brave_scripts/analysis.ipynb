```python
import json

def getLabelNames(instance):
    plain_list = []
    labels_list = json.loads(instance['labels'])
    for label in labels_list:
        plain_list.append(label['name'])
    return plain_list

def getLabelCounts(df):
    df = df.reset_index()
    label_counts = {}
    for index, row in df.iterrows():
        if row['invalidity'] != "invalid":
            continue
        label_list = getLabelNames(row)
        for label in label_list:
            if label in label_counts:
                label_counts[label] += 1
            else:
                label_counts[label] = 1

    return label_counts
```

## Retrieving the Invalid Bug Reports Without The _closed/duplicate_ label

```python
invalid_df = df[df['invalidity'] == 'invalid']
```

```python
invalid_no_duplicate = invalid_df[
    ~invalid_df.apply(lambda row: 'closed/duplicate' in getLabelNames(row), axis=1)
]

len(invalid_no_duplicate)
```

```python
# Sample 366 instances where invalidity == 'invalid' and no duplicate labeled bug reports
print(f"Total 'invalid' rows available: {len(invalid_no_duplicate)}")

if len(invalid_no_duplicate) < 366:
    raise ValueError(f"Not enough 'invalid' instances to sample 366 (found {len(invalid_no_duplicate)})")

sampled_invalid = invalid_no_duplicate.sample(n=366, random_state=42)
print("Sampled 366 'invalid' instances — showing the first 10:")
display(sampled_invalid.head(10))

# Save to CSV for later use
out_path = 'invalid_sample_no_duplicate_366.csv'
sampled_invalid.to_csv(out_path, index=False)
print(f"Saved sampled rows to {out_path}")
```

### Retrieve 5 samples from each invalid subclass on the 366 sized dataset 

```python
import pandas as pd

def sample_5_fill_false(g, n=5, seed=42):
    g_true = g[g["no_code_fix_present"] == True]
    g_false = g[g["no_code_fix_present"] == False]

    # Step 1: take as many True as possible (up to n)
    take_true = (
        g_true.sample(n=min(n, len(g_true)), random_state=seed)
        if len(g_true) > 0
        else g_true
    )

    remaining = n - len(take_true)

    # Step 2: fill remaining slots with False
    if remaining > 0 and len(g_false) > 0:
        take_false = g_false.sample(
            n=min(remaining, len(g_false)),
            random_state=seed
        )
        result = pd.concat([take_true, take_false], ignore_index=True)
    else:
        result = take_true

    return result.reset_index(drop=True)

samples_filled = (
    df_gemini
    .groupby("invalid_root_cause", group_keys=False)
    .apply(sample_5_fill_false)
    .reset_index(drop=True)
)

samples_filled.to_csv("invalid_gemini_sample.csv", index=False)

samples_filled.to_excel(
    "invalid_gemini_sample.xlsx",
    index=False,
    engine="openpyxl"   # or remove engine if openpyxl is default
)
```
